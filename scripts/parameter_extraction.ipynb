# Compute median value of parameters from participant

```python
import pandas as pd
```

```python
df_language = pd.read_csv("../data/data_participants_params_arbitration_language.csv")
df_observation = pd.read_csv("../data/data_participants_params_arbitration_observation.csv")
df_language.head()
```

```python
df_language["condition"] = "language"
df_observation["condition"] = "observation"
df = pd.concat([df_language, df_observation], ignore_index=True)
df.head()
```

```python
condition_median = df.groupby("condition").median(numeric_only=True)
overall_median = df.median(numeric_only=True)
combined = pd.concat([condition_median, overall_median.to_frame().T])
combined.index = list(condition_median.index) + ["overall"]
combined
```
