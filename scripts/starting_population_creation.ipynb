```python
import pandas as pd
```

```python
df_behaviour = pd.read_csv("../data/data_participants.csv")
df_behaviour["model_knowledge"] = df_behaviour["pl2_understanding"]

df_w_language = pd.read_csv("../data/data_participants_w_language.csv")
df_w_observation = pd.read_csv("../data/data_participants_w_observation.csv")
df_w_language["condition"] = "language"
df_w_observation["condition"] = "observation"
df_w = pd.concat([df_w_language, df_w_observation], ignore_index=True)

df = pd.DataFrame({
    "model_knowledge": df_behaviour[df_behaviour["gen"] == 0]["model_knowledge"]
        .round(2)
        .reset_index(drop=True),
    "w": ((df_w[df_w["generation"] == 1]["w"] * 100 - 50) / 50)
        .round(2)
        .reset_index(drop=True),
    "perf": df_behaviour[df_behaviour["gen"] == 0]["score_corrected"]
})

df.to_csv("../data/starting_population.csv", index=False)
```

Behavioural data

```python
df_behaviour = pd.read_csv("../data/data_participants.csv")
df_behaviour = (
    df_behaviour
    .query("gen == 0")
    [["ID", "score_corrected", "pl2_understanding"]]
    .rename(
        columns = {
        "pl2_understanding": "model_knowledge",
        "score_corrected": "perf"
        }
    )
)

df_behaviour.head()
```

W data

```python
df_w_language = pd.read_csv("../data/data_participants_w_language.csv")
df_w_observation = pd.read_csv("../data/data_participants_w_observation.csv")
df_w_language["condition"] = "language"
df_w_observation["condition"] = "observation"
df_w = pd.concat([df_w_language, df_w_observation], ignore_index=True)
df_w = (
    df_w
    .query("generation == 1")
    [["ID", "w"]]
    .assign(w = (df_w["w"] * 100 - 50) / 50)
)

df_w.head()
```

Merge dataframes

```python
df_merged = (
    df_behaviour
    .merge(df_w, on=["ID"], how="inner")
    [["ID", "perf", "model_knowledge", "w"]]
)

df_merged.head()
```

Save data

```python
df_merged.to_csv("../data/starting_population.csv", index=False)
```
