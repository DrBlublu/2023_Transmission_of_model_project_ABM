```python
import numpy as np

def create_transition_function(score: float) -> np.ndarray:
    """
    Create a (4,2,2) transition function with varied blocks
    but overall target score.
    """
    # Convert target score to average diagonal probability
    target_p = score / 100.0
    
    # Sample random diagonal values around target_p
    p_values = np.random.uniform(0, 1, size=4)
    # Rescale so their mean equals target_p
    p_values = p_values * (target_p / p_values.mean())
    p_values = np.clip(p_values, 0, 1)  # keep valid probabilities
    
    transition_function = np.zeros((4, 2, 2))
    for i, p in enumerate(p_values):
        transition_function[i] = np.array([
            [p, 1-p],
            [1-p, p]
        ])
    return transition_function

def compute_score(transition_function: np.ndarray) -> float:
    """
    Compute the score of a transition function.
    """
    return (transition_function[:,0,0] + transition_function[:,1,1]).mean() / 2 * 100


# Example usage
tf = create_transition_function(20)
print(tf)
print("Score:", compute_score(tf))
```
